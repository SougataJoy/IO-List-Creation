```python
from openpyxl import Workbook
# Import Workbook class
# Used to CREATE a new Excel file (.xlsx)
# Think: wb = Workbook() → creates blank Excel

from openpyxl.styles import Font, Alignment
# Import 2 styling tools:
# Font → controls text style (size, bold, color, font type)
# Alignment → controls text position inside cell (left, center, right, top, bottom)

from datetime import datetime
# Import datetime module
# Used to get current date & time from system
# Example: datetime.now() → current date/time

from openpyxl.styles import PatternFill
# Import PatternFill class
# Used to apply background color (fill) inside cells
# Like using "Fill Color" in Excel

from openpyxl.drawing.image import Image
# Import Image class
# Used to insert images (like logo.png) into Excel sheet
# Example: ws.add_image(img, "B2")

from google.colab import files
# Import Colab file utility
# Used to:
# - Upload files from your PC to Colab
# - Download files from Colab to your PC
# Example: files.upload(), files.download()

from openpyxl.styles import Border, Side
# Import border tools
# Side → defines one side of a border (thin, medium, thick)
# Border → combines 4 sides (left, right, top, bottom) into full border

from datetime import datetime
# Import datetime class from datetime module
# Used to work with date and time
# Example:
# datetime.now() → gives current system date & time

import pytz
# Import pytz library
# Used to handle timezones properly (like IST, UTC, etc.)
# Required because default datetime uses system/server time (Colab = UTC)
# Example:
# pytz.timezone('Asia/Kolkata') → sets Indian Standard Time (IST)

from openpyxl.utils import get_column_letter
# Import function from openpyxl
# Used to convert column NUMBER → Excel column LETTER
# Example:
# 1 → 'A'
# 2 → 'B'
# 26 → 'Z'
# 27 → 'AA'
# 28 → 'AB'
# This is required because Excel columns are NOT simple alphabets after Z

from openpyxl.utils import range_boundaries
from IPython.display import display

import os
# Import os module
# Used to interact with the operating system (OS)
# Allows you to:
# - List files and folders → os.listdir()
# - Delete files → os.remove()
# - Check if file exists → os.path.exists()
# - Work with file paths → os.path.join()
# - Create/remove directories → os.mkdir(), os.rmdir()

# In Google Colab:
# Used to manage uploaded and generated files in the runtime environment

# Example:
# os.listdir() → shows all files in current folder
# os.remove("file.xlsx") → deletes a file

for file in os.listdir():                     #Remove command: for last uploads
    if file.endswith(".png") or file.endswith(".xlsx"):
        os.remove(file)

thin = Side(style='thin')   # define thin border line

# Upload file
uploaded = files.upload()

# Automatically get filename
logo_file = list(uploaded.keys())[0]

# Load image
logo = Image(logo_file)

# Set timezone to India
ist = pytz.timezone('Asia/Kolkata')

# Get current IST time
today_date = datetime.now(ist).strftime("%Y-%b-%d_%H-%M")
today_date_only = datetime.now(ist).strftime("%d-%m-%Y")

# Create workbook
wb = Workbook()

# Create CoverPage sheet
ws = wb.active
ws.title = "CoverPage"

#Set dimensions
# ---- SETTINGS ----
col_width = 5.7
row_height = 25

# Turn off gridlines
ws.sheet_view.showGridLines = False

# ---- APPLY TO COLUMNS (A to S) ----
for col in range(1, 41):  # 1=A, 39=AN
    col_letter = get_column_letter(col)
    ws.column_dimensions[col_letter].width = col_width
# ---- APPLY TO ROWS ----
for row in range(1, 50):  # adjust range as needed
    ws.row_dimensions[row].height = row_height

# Add to sheet
ws.add_image(logo, "B2")

# Resize
logo.width = 84
logo.height = 69

border_thin_Left = Border(left=thin, top=thin, bottom=thin)
# Used for LEFT EDGE of a table
# Applies:
# - left border (outer edge)
# - top & bottom (to continue row lines)

border_thin_Top_Bottom = Border(top=thin, bottom=thin)
# Used for MIDDLE cells
# Applies:
# - only horizontal lines (top & bottom)
# - no left/right (so no double lines)

border_thin_Right = Border(right=thin, top=thin, bottom=thin)
# Used for RIGHT EDGE of a table
# Applies:
# - right border (outer edge)
# - top & bottom (to continue row lines)

border_All = Border(left=thin, right=thin, top=thin, bottom=thin)
# Used for FULL BORDER around a cell
# Applies:
# - left border (outer/inner vertical line)
# - right border (outer/inner vertical line)
# - top border (horizontal line)
# - bottom border (horizontal line)
# Creates a complete box around the cell (all 4 sides)

# Create Border for 1st row
ws.cell(row=6, column=3).border = border_thin_Left   # C6
ws.cell(row=6, column=38).border = border_thin_Top_Bottom  # AL6
ws.cell(row=6, column=39).border = border_thin_Right  # AM6
ws.cell(row=6, column=21).border = border_thin_Top_Bottom  # U6
ws.cell(row=6, column=24).border = border_thin_Top_Bottom  # X6
ws.cell(row=6, column=36).border = border_thin_Left  # AJ6

# Merge cells for for 1st Row
ws.merge_cells('C6:T6')
ws.merge_cells('U6:W6')
ws.merge_cells('X6:AI6')
ws.merge_cells('AJ6:AK6')
ws.merge_cells('AM6:AN6')
for row in range(6, 18):   # 18 because range stops at 17
    ws[f'C{row}'].font = Font(name='Calibri', size=18, bold=True)
    ws[f'C{row}'].alignment = Alignment(horizontal='right')
    ws[f'U{row}'].font = Font(name='Calibri', size=18, bold=True)
    ws[f'U{row}'].alignment = Alignment(horizontal='center')
    ws[f'X{row}'].font = Font(name='Calibri', size=16, bold=False)
    ws[f'X{row}'].alignment = Alignment(horizontal='left')

ws['AJ6'].font = Font(name='Calibri', size=18, bold=True)
ws['AJ6'].alignment = Alignment(horizontal='right')
ws['AL6'].font = Font(name='Calibri', size=18, bold=True)
ws['AL6'].alignment = Alignment(horizontal='center')
ws['AM6'].font = Font(name='Calibri', size=16, bold=False)
ws['AM6'].alignment = Alignment(horizontal='left')



# Apply repeating merges
for row in range(7, 17):
    ws.merge_cells(f'C{row}:T{row}')
#for row in range(6, 17):
    ws.merge_cells(f'U{row}:W{row}')
#for row in range(6, 17):
    ws.merge_cells(f'X{row}:AN{row}')
    ws.cell(row=row, column=3).border = border_thin_Left
    ws.cell(row=row, column=40).border = border_thin_Right
# --- C to T (3 to 20) ---
# --- C to AN (3 to 40) ---
    for col in range(4, 40):
        ws.cell(row=row, column=col).border = border_thin_Top_Bottom
# --- U to W (21 to 23) ---
#for col in range(21, 24):
        ws.cell(row=row, column=col).border = border_thin_Top_Bottom

ws.merge_cells('AJ6:AK6')
ws.merge_cells('AM6:AN6')

# Apply repeating merges again
from openpyxl.utils import get_column_letter

# Define your column blocks
blocks = [
    (3, 6),   # C–F
    (7, 14),  # G–N
    (15, 22),  # O–V
    (23, 28),  # W–AB
    (29, 34),  # AC–AH
    (35, 40),  # AI–AN
    ]

for row in range(23, 29):
    for start_col, end_col in blocks:

        # Convert column numbers → letters
        start_letter = get_column_letter(start_col)
        end_letter = get_column_letter(end_col)

        # Merge cells
        ws.merge_cells(f'{start_letter}{row}:{end_letter}{row}')

        # Left border
        ws.cell(row=row, column=start_col).border = border_thin_Left

        # Right border
        ws.cell(row=row, column=end_col).border = border_thin_Right

        # Middle cells (Top & Bottom)
        for col in range(start_col + 1, end_col):
            ws.cell(row=row, column=col).border = border_thin_Top_Bottom


def apply_outer_border(ws, cell_range):
    thick = Side(style='thick')

    min_col, min_row, max_col, max_row = range_boundaries(cell_range)

    for row in range(min_row, max_row + 1):
        for col in range(min_col, max_col + 1):
            cell = ws.cell(row=row, column=col)

            b = cell.border  # existing border

            cell.border = Border(
                top=thick if row == min_row else b.top,
                bottom=thick if row == max_row else b.bottom,
                left=thick if col == min_col else b.left,
                right=thick if col == max_col else b.right,
            )

#Apply thick border
apply_outer_border(ws, "C23:AN28")

# Add Left side Title Row 1
ws['C6'] = "DWG/DOC TITLE"
ws['C7'] = "PROJECT NAME"
ws['C8'] = "CLIENT"
ws['C9'] = "END USER"
ws['C10'] = "VENDOR"
ws['C11'] = "PO REFERENCE"
ws['C12'] = "JC NO"
ws['C13'] = "PACKAGE NAME"
ws['C14'] = "REFERENCE CLIENT DOC. NAME / NO"
ws['C15'] = "REFERENCE VENDOR DOCUMENT NO"
ws['C16'] = "CUSTOMER PROJECT CODE"


# Add Middle side Title Row 2
for row in range(6, 17):   # 18 because range stops at 16
    ws[f'U{row}'] = ":"

# ---- USER INPUT ----
#-- & --
# Add Middle side Title Row 3 & rest
ws['X6'] = "IO List"
ws['AJ6'] = "REV."
ws['AL6'] = ":"
ws['AM6'] = "R0"
project_name = input("Enter Project Name: ").title()
ws['X7'] = project_name
client = input("Enter Client Name: M/s. ").title()
ws['X8'] = f"M/s. {client}"
end_user = input("Enter End User Name: M/s. ").title()
ws['X9'] = f"M/s. {end_user}"
ws['X10'] = "M/s. Subtleweigh Electric (I) Pvt. Ltd."
PO_ref = input("Enter PO Reference No.").upper()
ws['X11'] = PO_ref
JC_no = input("Enter JC No.").title()
ws['X12'] = JC_no
Package_Nm = input("Enter Package Name").title()
ws['X13'] = Package_Nm
Ref_Client_Doc = input("Enter Reference Client Doc").title()
ws['X14'] = Ref_Client_Doc
Ref_Vendor_Doc = input("Enter Reference Vendor Doc").title()
ws['X15'] = Ref_Vendor_Doc
Cust_Proj_Code = input("Enter Customer Proj. Code").title()
ws['X16'] = Cust_Proj_Code
engineer = input("Enter Engineer Name: ")
ws['C28'] = "REV. NO."
ws['G28'] = "DATE"
ws['O28'] = "DESCRIPTION"
ws['W28'] = "PREPARED BY"
ws['AC28'] = "CHECKED BY"
ws['AI28'] = "APPROVED BY"
ws['C27'] = "R0"
ws['G27'] = today_date_only
ws['O27'] = "FOR APPROVAL"
ws['W27'] = "SP"
ws['AC27'] = "SR"
ws['AI27'] = "PKR"
ws['C28'].font = Font(name='Calibri', size=18, bold=True)
ws['C28'].alignment = Alignment(horizontal='right')
ws['G28'].font = Font(name='Calibri', size=18, bold=True)
ws['G28'].alignment = Alignment(horizontal='center')
ws['O28'].font = Font(name='Calibri', size=18, bold=True)
ws['O28'].alignment = Alignment(horizontal='center')
ws['W28'].font = Font(name='Calibri', size=18, bold=True)
ws['W28'].alignment = Alignment(horizontal='center')
ws['AC28'].font = Font(name='Calibri', size=18, bold=True)
ws['AC28'].alignment = Alignment(horizontal='center')
ws['AI28'].font = Font(name='Calibri', size=18, bold=True)
ws['AI28'].alignment = Alignment(horizontal='center')

for col in range(start_col, end_col + 1):
    col_letter = chr(64 + col)  # convert number → letter

    ws[f'{col_letter}6'].font = Font(name='Calibri', size=18, bold=True)
    ws[f'{col_letter}6'].alignment = Alignment(horizontal='center')

# 👉 Add new sheet
sheet_name = input("Enter sheet name: ")
if sheet_name in wb.sheetnames:
    print("Sheet already exists!")
else:
    wb.create_sheet(title=sheet_name)
# ---- APPLY TO COLUMNS (A to S) ----
for col in range(1, 40):  # 1=A, 39=AN
    col_letter = get_column_letter(col)
    ws.column_dimensions[col_letter].width = col_width
# ---- APPLY TO ROWS ----
for row in range(1, 50):  # adjust range as needed
    ws.row_dimensions[row].height = row_height

# Save file
file_name = f"{today_date}_IO_List.xlsx"
wb.save(file_name)

print(f"{today_date} IO List created")
```

```python
from google.colab.output import disable_custom_widget_manager
import ipywidgets as widgets
from IPython.display import display

node = widgets.IntText(description="Node:")
slot = widgets.IntText(description="Slot:")
card = widgets.Dropdown(
    options=["AI", "DI", "DO", "AO"],
    description="Type:"
)
qty = widgets.IntText(description="Qty:")

display(node, slot, card, qty)

data = []

button = widgets.Button(description="Add")
output = widgets.Output()

def on_click(b):
    data.append((node.value, slot.value, card.value, qty.value))
    with output:
        print(f"Added: N{node.value} S{slot.value} {card.value} x{qty.value}")

button.on_click(on_click)

display(button, output)
print(data)
```
