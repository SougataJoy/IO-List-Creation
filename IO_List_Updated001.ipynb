```python
from openpyxl import Workbook
from openpyxl.styles import Font, Alignment
from datetime import datetime
from openpyxl.styles import PatternFill
from openpyxl.drawing.image import Image
from google.colab import files
from openpyxl.styles import Border, Side
from openpyxl.utils import range_boundaries
from IPython.display import display
import os
import pytz
from openpyxl.utils import get_column_letter
# Import function from openpyxl
# Used to convert column NUMBER → Excel column LETTER
# Example:
# 1'A',2'B',3'C',4'D',5'E',6'F',7'G',8"H",9'I'......26'Z'
# 27'AA',28'AB',29'AC',30'AD',31'AE',32'AF',33'AG',33"AH",34'AI'......52'AZ'
# This is required because Excel columns are NOT simple alphabets after Z


for file in os.listdir():   #Required Only in Colab Remove command: for last uploads
    if file.endswith(".png") or file.endswith(".xlsx"):
        os.remove(file)

thin = Side(style='thin')   # define thin border line

# Upload file
uploaded = files.upload()

# Automatically get filename
logo_file = list(uploaded.keys())[0]

# Load image
logo = Image(logo_file)

# Set timezone to India
ist = pytz.timezone('Asia/Kolkata')

# Get current IST time
today_date = datetime.now(ist).strftime("%Y-%b-%d_%H-%M")
today_date_only = datetime.now(ist).strftime("%d-%m-%Y")

# Create workbook
wb = Workbook()

# Create CoverPage sheet
ws = wb.active
ws.title = "CoverPage"

#Set dimensions
# ---- SETTINGS ----
col_width = 5.7
row_height = 25

# Turn off gridlines
ws.sheet_view.showGridLines = False
def apply_border_thin(ws, cell_range):

    min_col, min_row, max_col, max_row = range_boundaries(cell_range)

    for row in range(min_row, max_row + 1):
        for col in range(min_col, max_col + 1):
            cell = ws.cell(row=row, column=col)

            b = cell.border  # existing border

            cell.border = Border(
                top=thin if row == min_row else b.top,
                bottom=thin if row == max_row else b.bottom,
                left=thin if col == min_col else b.left,
                right=thin if col == max_col else b.right,
            )

# ---- APPLY TO COLUMNS (A to S) ----
for col in range(1, 41):  # 1=A, 39=AN
    col_letter = get_column_letter(col)
    ws.column_dimensions[col_letter].width = col_width
# ---- APPLY TO ROWS ----
for row in range(1, 50):  # adjust range as needed
    ws.row_dimensions[row].height = row_height

# Add to sheet
ws.add_image(logo, "B2")

# Resize
logo.width = 84
logo.height = 69

# Merge cells for for 1st Row
ws.merge_cells('C6:T6')
ws.merge_cells('U6:W6')
ws.merge_cells('X6:AI6')
ws.merge_cells('AJ6:AK6')
ws.merge_cells('AM6:AN6')

def merge_cells_block(ws, start_col, end_col, row):
    # Merge
    ws.merge_cells(start_row=row, start_column=start_col,
                   end_row=row, end_column=end_col)

for row in range(7, 17):
    merge_cells_block(ws, 3, 20, row)   # C–T
    merge_cells_block(ws, 21, 23, row)  # U–W
    merge_cells_block(ws, 24, 40, row)  # X–AN

ws['AJ6'].font = Font(name='Calibri', size=18, bold=True)
ws['AJ6'].alignment = Alignment(horizontal='right')
ws['AL6'].font = Font(name='Calibri', size=18, bold=True)
ws['AL6'].alignment = Alignment(horizontal='center')
ws['AM6'].font = Font(name='Calibri', size=16, bold=False)
ws['AM6'].alignment = Alignment(horizontal='left')

def style_cell(ws, row, col, font=None, align=None, border=None, value=None):
    cell = ws.cell(row=row, column=col)

    if value is not None:
        cell.value = value
    if font:
        cell.font = font
    if align:
        cell.alignment = align
    if border:
        cell.border = border

for row in range(6, 17):   # 17 because range stops at 16
    style_cell(ws, row, 3,
              font=Font(name='Calibri', size=18, bold=True),
              align=Alignment(horizontal='right'),
               )
    style_cell(ws, row, 21,
              font=Font(name='Calibri', size=16, bold=True),
              align=Alignment(horizontal='center'),
              )
    ws[f'X{row}'].font = Font(name='Calibri', size=16, bold=False)
    ws[f'X{row}'].alignment = Alignment(horizontal='left')

# Define your column blocks
blocks = [
    (3, 6),   # C–F
    (7, 14),  # G–N
    (15, 22),  # O–V
    (23, 28),  # W–AB
    (29, 34),  # AC–AH
    (35, 40),  # AI–AN
    ]

for row in range(23, 29):                 # Upto row 28
    for start_col, end_col in blocks:

        # Convert column numbers → letters
        start_letter = get_column_letter(start_col)
        end_letter = get_column_letter(end_col)

        # Merge cells
        ws.merge_cells(f'{start_letter}{row}:{end_letter}{row}')

    apply_border_thin(ws, f"C{row}:F{row}")
    apply_border_thin(ws, f"G{row}:N{row}")
    apply_border_thin(ws, f"O{row}:V{row}")
    apply_border_thin(ws, f"W{row}:AB{row}")
    apply_border_thin(ws, f"AC{row}:AH{row}")
    apply_border_thin(ws, f"AI{row}:AN{row}")

#Apply thin border
apply_border_thin(ws, "AJ6:AN6")
for row in range(6, 17):   # Upto row 16
    apply_border_thin(ws, f"C{row}:AN{row}")

def apply_outer_border(ws, cell_range):
    thick = Side(style='thick')

    min_col, min_row, max_col, max_row = range_boundaries(cell_range)

    for row in range(min_row, max_row + 1):
        for col in range(min_col, max_col + 1):
            cell = ws.cell(row=row, column=col)

            b = cell.border  # existing border

            cell.border = Border(
                top=thick if row == min_row else b.top,
                bottom=thick if row == max_row else b.bottom,
                left=thick if col == min_col else b.left,
                right=thick if col == max_col else b.right,
            )

#Apply thick border
apply_outer_border(ws, "C23:AN28")

# Add Left side Title Row 1
labels = {
    6: "DWG/DOC TITLE",
    7: "PROJECT NAME",
    8: "CLIENT",
    9: "END USER",
    10: "VENDOR",
    11: "PO REFERENCE",
    12: "JC NO",
    13: "PACKAGE NAME",
    14: "REFERENCE CLIENT DOC. NAME / NO",
    15: "REFERENCE VENDOR DOCUMENT NO",
    16: "CUSTOMER PROJECT CODE"
}

for row, text in labels.items():
    ws[f'C{row}'] = text

# Add Middle side Title Row 2
for row in range(6, 17):   # 18 because range stops at 16
    ws[f'U{row}'] = ":"

# ---- USER INPUT ----
#-- & --
# Add Middle side Title Row 3 & rest
ws['X6'] = "IO List"
ws['AJ6'] = "REV."
ws['AL6'] = ":"
ws['AM6'] = "R0"
ws['X10'] = "M/s. Subtleweigh Electric (I) Pvt. Ltd."

inputs = [
   ("X7", "Enter Project Name", "title"),
   ("X8", "Enter Client Name", "title", "M/s. "),
   ("X9", "Enter End User Name", "title", "M/s. "),
   ("X11", "Enter PO Reference No.", "upper"),
   ("X12", "Enter JC No.", "upper"),
   ("X13", "Enter Package Name", "title"),
   ("X14", "Enter Reference Client Doc", "title"),
   ("X15", "Enter Reference Vendor Doc", "title"),
   ("X16", "Enter Customer Proj. Code", "title"),
]

for item in inputs:
    cell, prompt, case = item[:3]
    prefix = item[3] if len(item) > 3 else ""

    value = input(prompt + ": ").strip()
    if not value:
        value = "N/A"
    if case == "title":
        value = value.title()
    elif case == "upper":
        value = value.upper()

    ws[cell] = f"{prefix}{value}"

ws['C28'] = "REV. NO."
ws['G28'] = "DATE"
ws['O28'] = "DESCRIPTION"
ws['W28'] = "PREPARED BY"
ws['AC28'] = "CHECKED BY"
ws['AI28'] = "APPROVED BY"
ws['C27'] = "R0"
ws['G27'] = today_date_only
ws['O27'] = "FOR APPROVAL"
ws['W27'] = "SP"
ws['AC27'] = "SR"
ws['AI27'] = "PKR"

cols = ['C', 'G', 'O', 'W', 'AC', 'AI']

for row, size, bold in [(28, 18, True), (27, 16, False)]:
    for col in cols:
        cell = ws[f'{col}{row}']
        cell.font = Font(name='Calibri', size=size, bold=bold)
        cell.alignment = Alignment(horizontal='center'
        )

# 👉 Add new sheet
sheetQty = int(input("Enter Qty of sheets: "))

try:
    for i in range(sheetQty):
        print(f"Total Qty of page/s: {i}")
except ValueError:
    print("Please enter a valid number")

def add_new_sheet(ws):
  sheet_name = input("Enter sheet name: ")
  if sheet_name in wb.sheetnames:
      print("Sheet already exists!")
  else:
      ws01 = wb.create_sheet(title=sheet_name)
  # Create sheet
  ws = wb.active
  ws.title = sheet_name


#for i in range(1, int(sheetQty) + 1):

# Save file
file_name = f"{today_date}_IO_List.xlsx"
wb.save(file_name)

print(f"{today_date} IO List created")
```
