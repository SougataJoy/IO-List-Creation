```python
from openpyxl import Workbook
# Creates a new Excel workbook (.xlsx file)

from openpyxl.styles import Font, Alignment
# Font → controls text style (size, bold, color, font name)
# Alignment → controls text alignment (horizontal, vertical, wrap text, etc.)

from openpyxl.styles import PatternFill
# Applies background color to cells (like Excel "Fill Color")

from openpyxl.styles import Border, Side
# Side → defines border style for one side (left/right/top/bottom: thin, thick, color)
# Border → combines multiple sides to apply full border to a cell

from openpyxl.drawing.image import Image
# Inserts images (e.g., company logo) into Excel sheet

from datetime import datetime
# Handles date & time (current time, formatting, calculations, timestamps)

from openpyxl.utils import get_column_letter
# Converts column number → Excel column letter (1→A, 27→AA)

from openpyxl.utils import range_boundaries
# Converts Excel range (e.g., "C6:T17") → numeric limits (min_col, min_row, max_col, max_row)

import pytz
# Handles timezone-aware datetime (e.g., IST, UTC conversion)

import os
# Used for file/folder operations (list, delete, rename files)

# Dedicated only for Jupyter / Google Colab
from IPython.display import display
# Displays rich/pretty output (tables, DataFrames, objects) in notebook

from google.colab import files
# Upload/download files between your local PC and Google Colab

#Remove command: for last uploads
for file in os.listdir():
    if file.endswith(".png") or file.endswith(".xlsx"):
        os.remove(file)

# Upload file
uploaded = files.upload()

# Automatically get filename
logo_file = list(uploaded.keys())[0]

# Load image
logo = Image(logo_file)

# Set timezone to India
ist = pytz.timezone('Asia/Kolkata')

# Get current IST time
today_date = datetime.now(ist).strftime("%Y-%b-%d_%H-%M")
today_date_only = datetime.now(ist).strftime("%d-%m-%Y")


# Create workbook
wb = Workbook()

# Create CoverPage sheet
ws = wb.active
ws.title = "CoverPage"

# Turn off gridlines
ws.sheet_view.showGridLines = False

thin = Side(style='thin')     # define thin border line
thick = Side(style='thick')   # define thick border line

border_thin_Left = Border(left=thin, top=thin, bottom=thin)
# Used for LEFT EDGE of a table
border_thin_Top_Bottom = Border(top=thin, bottom=thin)
# Used for MIDDLE cells
border_thin_Right = Border(right=thin, top=thin, bottom=thin)
# Used for RIGHT EDGE of a table
border_All = Border(left=thin, right=thin, top=thin, bottom=thin)
# Used for FULL BORDER around a cell

# Define your column blocks
blocks = [
    (3, 20, 20, 24),   # C to T, rows 20 to 24
]

for start_col, end_col, start_row, end_row in blocks:

    # Convert column numbers → letters
    start_letter = get_column_letter(start_col)
    end_letter = get_column_letter(end_col)

    # Merge cells
    ws.merge_cells(f'{start_letter}{start_row}:{end_letter}{end_row}')

    # Apply borders
    for row in range(start_row, end_row + 1):
        for col in range(start_col, end_col + 1):

            cell = ws.cell(row=row, column=col)

            # Left border
            if col == start_col:
                cell.border = border_thin_Left

            # Top & Bottom borders
            elif start_row <= row <= end_row:
                cell.border = border_thin_Top_Bottom

# Save file
file_name = f"{today_date}_IO_List.xlsx"
wb.save(file_name)

print(f"{today_date} IO List created")
```
